```python
#Stemmer

# stemmer is a tool in NLP that reduces a word to its base/root form
# removes suffixes like:-ing,-ed,-ly etc

from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = ["playing", "played", "studies", "happiness"]

for word in words:
    print(word, "→", stemmer.stem(word))
```

```python
# lemmatization  -> Lemmatization is the process of reducing a word to its base or dictionary form, called a lemma.
import spacy
from nltk.corpus import wordnet
import language_tool_python
import nltk

# Load tools
nlp = spacy.load("en_core_web_sm")
tool = language_tool_python.LanguageTool('en-US')

def noun_to_adj(word):
    """
    Enhanced: Checks for pertainyms AND derivationally related adjectives.
    """
    syns = wordnet.synsets(word.lower(), pos=wordnet.NOUN)
    for s in syns:
        for lemma in s.lemmas():
            # 1. Your original Pertainym check
            for pertainym in lemma.pertainyms():
                return pertainym.name().replace("_", " ")
            
            # 2. NEW: Check for derivationally related forms that are Adjectives
            for drf in lemma.derivationally_related_forms():
                if drf.synset().pos() in ['a', 's']: # 'a' = adj, 's' = satellite adj
                    return drf.name().replace("_", " ")
    return word

def smart_lemmatize(text):
    doc = nlp(text)
    tokens = []
    for token in doc:
        if token.pos_ == "NOUN":
            # Pass the lemma (e.g., 'happiness') to the converter
            lemma = noun_to_adj(token.lemma_)
        else:
            lemma = token.lemma_
        tokens.append(lemma)
    
    lemmatized_text = " ".join(tokens)
    matches = tool.check(lemmatized_text)
    return language_tool_python.utils.correct(lemmatized_text, matches)

text = "He was feeling happiness"
print(smart_lemmatize(text))
```

```python
# Stemming : Rule-based chopping ,fast
# Lemmatization : Dictionary + POS + grammar aware ,slow
```
